# Pipeline Vitis AI para ZCU104

Este notebook monta e documenta os comandos. A execucao oficial deve ser feita pelos scripts no terminal do container Vitis AI.

```python
MODEL = 'depth_reduced'  # baseline, depth_reduced, skip_connections, mobilenet_v2, mobilenet_v3
TARGET = 'DPUCZDX8G_ISA1_B4096'  # ajuste ao target real da sua DPU
CSV_CALIB = 'STARCOP_mini/train_mini10.csv'
CSV_TEST = 'STARCOP_mini/test_mini10.csv'
DATA_ROOT = 'STARCOP_mini'
ARCH_JSON = '/caminho/exato/para/arch.json'
```

```python
commands = {
    'inspect': f'python VitisAI/inspect_model.py --model {MODEL} --target {TARGET}',
    'calib': f'python VitisAI/quantize_model.py --model {MODEL} --quant-mode calib --csv {CSV_CALIB} --data-root {DATA_ROOT} --subset-len 100 --target {TARGET}',
    'export': f'python VitisAI/quantize_model.py --model {MODEL} --quant-mode test --csv {CSV_TEST} --data-root {DATA_ROOT} --subset-len 1 --target {TARGET} --deploy',
}
for stage, command in commands.items():
    print(f'[{stage}]\n{command}\n')
```

Copie os comandos impressos e execute-os, na ordem, dentro do Docker. Depois localize o `*_int.xmodel` e compile usando:

`python VitisAI/compile_xmodel.py --xmodel <arquivo_int.xmodel> --arch <arch.json> --name methane_<modelo>`
